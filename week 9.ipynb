```python
import pandas as pd
from sklearn.tree import DecisionTreeRegressor

# 1. Load the dataset
data = pd.read_csv('acc.csv')

# 2. Extract features (X: columns from index 3 onward) and target variable (y: index 2 'freq')
X = data.iloc[:, 3:]  # Features: ['dead', 'casu', 'serious', 'light', 'report']
y = data.iloc[:, 2]   # Target: 'freq'

# Exclude the last row (Row 15) from the training set to use 0 to 14 for training
X_train = X.iloc[:-1, :]
y_train = y.iloc[:-1]

# 3. Initialize and train the DecisionTreeRegressor model
regr = DecisionTreeRegressor(random_state=42)
regr.fit(X_train, y_train)

# 4. Predict accident frequency for the excluded unseen test data (Row 15)
# Input features: dead=202, casu=11008, serious=3846, light=6770, report=392
new_data = pd.DataFrame(
    [[202, 11008, 3846, 6770, 392]],
    columns=X_train.columns
)
y_pred = regr.predict(new_data)

# 5. Display prediction results and compare with the ground truth
actual_y = y.iloc[-1]
print("=== [Program 2: Predicting Accident Frequency (freq)] ===")
print(f"• Predicted Frequency: {y_pred[0]:.0f}")
print(f"• Actual Frequency   : {actual_y:.0f}")
print(f"• Absolute Error     : {abs(actual_y - y_pred[0]):.0f}")
```

```python

import pandas as pd
from sklearn.tree import DecisionTreeRegressor

# 1. Load the dataset
data = pd.read_csv('acc.csv')
data = data.dropna()

# 2. Define features (X) and target variable (y)
X = data[['Violation', 'day/night', 'dead']]
y = data['casu']

# Convert categorical features into numeric columns
X = pd.get_dummies(X, columns=['Violation', 'day/night'], dtype=int)

# Exclude the last row (Row 15) from the training set
X_train = X.iloc[:-1, :]
y_train = y.iloc[:-1]

# 3. Initialize and train the DecisionTreeRegressor model
regr = DecisionTreeRegressor(random_state=42)
regr.fit(X_train, y_train)

# 4. Predict casualties for the excluded test data (last row)
X_test = X.iloc[[-1], :]
y_pred = regr.predict(X_test)

# 5. Display prediction results and compare with the ground truth
actual_y = y.iloc[-1]

print("=== [Program 2: Predicting Casualties (casu)] ===")
print(f"• Predicted Casualties: {y_pred[0]:.0f}")
print(f"• Actual Casualties   : {actual_y:.0f}")
print(f"• Absolute Error      : {abs(actual_y - y_pred[0]):.0f}")
```

```python

import pandas as pd
from sklearn.tree import DecisionTreeRegressor

# 1. Load the dataset and remove missing values
data = pd.read_csv('acc.csv')
data = data.dropna()

# 2. Define features (X) and target variable (y)
X = data[['freq', 'casu', 'serious']]
y = data['light']

# 3. Exclude the last row from training
X_train = X.iloc[:-1, :]
y_train = y.iloc[:-1]

# 4. Use the last row for testing
X_test = X.iloc[[-1], :]
actual_y = y.iloc[-1]

# 5. Initialize and train the model
regr = DecisionTreeRegressor(random_state=42)
regr.fit(X_train, y_train)

# 6. Predict light injuries
y_pred = regr.predict(X_test)

# 7. Display prediction results and compare with actual value
print("=== Predicting Light Injuries (light) ===")
print(f"Predicted Light Injuries: {y_pred[0]:.0f}")
print(f"Actual Light Injuries   : {actual_y:.0f}")
print(f"Absolute Error          : {abs(actual_y - y_pred[0]):.0f}")
```
